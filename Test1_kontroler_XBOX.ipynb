```python
import ipywidgets as widgets
from jetracer.nvidia_racecar import NvidiaRacecar
import traitlets
from IPython.display import display

controller = widgets.Controller(index=0)
display(controller)

print("Kontroler wyświetlony. Porusz gałką i uruchom następną komórkę.")
```

```python
print("axes:", controller.axes)
print("buttons:", controller.buttons)
print("axes count:", len(controller.axes))
print("buttons count:", len(controller.buttons))
```

```python
car = NvidiaRacecar()
car.throttle_gain = 0.5
car.steering_offset = 0
car.steering_gain = -0.65
car.steering = 0
car.throttle = 0
```

```python
# skręt: lewa gałka lewo/prawo
steering_link = traitlets.dlink(
    (controller.axes[0], 'value'),
    (car, 'steering'),
    transform=lambda x: -x
)

# gaz: przycisk RT
throttle_link = traitlets.dlink(
    (controller.buttons[7], 'value'),
    (car, 'throttle'),
    transform=lambda x: -x
)

# hamulec/wsteczny: przycisk LT
throttle_link = traitlets.dlink(
    (controller.buttons[6], 'value'),
    (car, 'throttle'),
    transform=lambda x: x
)
    
```
